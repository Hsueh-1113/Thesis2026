```python
from pathlib import Path
from openpyxl import load_workbook

# 改成你的 out 資料夾路徑
out_dir = Path(r"/Users/lishixue/Documents/Master thesis/Statafile/Thesis2026/out/Tables")


# ---------- helpers ----------
def load_ws(xlsx_path: Path):
    wb = load_workbook(xlsx_path, data_only=True)
    return wb.active


def cell(ws, addr: str) -> str:
    v = ws[addr].value
    return "" if v is None else str(v)


def esc(s: str) -> str:
    s = "" if s is None else str(s)
    repl = {
        "&": r"\&",
        "%": r"\%",
        "$": r"\$",
        "#": r"\#",
        "_": r"\_",
    }
    for k, v in repl.items():
        s = s.replace(k, v)
    return s


def write_text(path: Path, text: str):
    path.write_text(text, encoding="utf-8")
    print(f"Saved: {path}")


def table_template(title: str, colspec: str, header: str, body: str, notes: str) -> str:
    return rf"""
\begin{{table}}[htbp]
\centering
\caption{{{title}}}
\small
\begin{{adjustbox}}{{max width=\textwidth}}
\begin{{tabular}}{{{colspec}}}
\toprule
{header}
\midrule
{body}
\bottomrule
\end{{tabular}}
\end{{adjustbox}}

\begin{{flushleft}}
\footnotesize {notes}
\end{{flushleft}}
\end{{table}}
""".strip()


# ---------- employment ----------
def convert_employment():
    xlsx = out_dir / "employment_tables" / "Table_BAME_custom.xlsx"
    tex = out_dir / "employment_tables" / "Table_BAME_custom.tex"
    ws = load_ws(xlsx)

    title = esc(cell(ws, "A2"))

    body = rf"""
\multicolumn{{8}}{{l}}{{\textit{{Change in employment status}}}} \\
No change & {esc(cell(ws, "D8"))} &  &  & {esc(cell(ws, "G8"))} &  &  & {esc(cell(ws, "J8"))} \\
& {esc(cell(ws, "D9"))} &  &  & {esc(cell(ws, "G9"))} &  &  & \\
Reduction & {esc(cell(ws, "D10"))} &  &  & {esc(cell(ws, "G10"))} &  &  & {esc(cell(ws, "J10"))} \\
& {esc(cell(ws, "D11"))} &  &  & {esc(cell(ws, "G11"))} &  &  & \\
Job loss & {esc(cell(ws, "D12"))} &  &  & {esc(cell(ws, "G12"))} &  &  & {esc(cell(ws, "J12"))} \\
& {esc(cell(ws, "D13"))} &  &  & {esc(cell(ws, "G13"))} &  &  & \\
\addlinespace
\multicolumn{{8}}{{l}}{{\textit{{Key worker in the baseline}}}} \\
Yes &  & {esc(cell(ws, "E18"))} &  &  & {esc(cell(ws, "H18"))} &  & {esc(cell(ws, "J18"))} \\
&  & {esc(cell(ws, "E19"))} &  &  & {esc(cell(ws, "H19"))} &  & \\
No &  & {esc(cell(ws, "E20"))} &  &  & {esc(cell(ws, "H20"))} &  & {esc(cell(ws, "J20"))} \\
&  & {esc(cell(ws, "E21"))} &  &  & {esc(cell(ws, "H21"))} &  & \\
\addlinespace
\multicolumn{{8}}{{l}}{{\textit{{Self-employed in the baseline}}}} \\
Yes &  &  & {esc(cell(ws, "F24"))} &  &  & {esc(cell(ws, "I24"))} & {esc(cell(ws, "J24"))} \\
&  &  & {esc(cell(ws, "F25"))} &  &  & {esc(cell(ws, "I25"))} & \\
No &  &  & {esc(cell(ws, "F26"))} &  &  & {esc(cell(ws, "I26"))} & {esc(cell(ws, "J26"))} \\
&  &  & {esc(cell(ws, "F27"))} &  &  & {esc(cell(ws, "I27"))} & \\
\midrule
Observations & {esc(cell(ws, "D29"))} & {esc(cell(ws, "E29"))} & {esc(cell(ws, "F29"))} & {esc(cell(ws, "G29"))} & {esc(cell(ws, "H29"))} & {esc(cell(ws, "I29"))} & \\
""".strip()

    header = r"& Bame & Bame & Bame & Non-Bame & Non-Bame & Non-Bame & p-value \\"
    notes = esc(cell(ws, "A30")) + " " + esc(cell(ws, "A31"))
    tex_text = table_template(title, r"p{4.8cm}ccccccc", header, body, notes)
    write_text(tex, tex_text)


# ---------- financial ----------
def convert_financial():
    xlsx = out_dir / "financial_tables" / "Table_BAME_financial.xlsx"
    tex = out_dir / "financial_tables" / "Table_BAME_financial.tex"
    ws = load_ws(xlsx)

    title = esc(cell(ws, "A2"))

    body = rf"""
\multicolumn{{6}}{{l}}{{\textit{{Financial difficulties at baseline}}}} \\
No (1-3) & {esc(cell(ws, "D8"))} &  & {esc(cell(ws, "F8"))} &  & {esc(cell(ws, "H8"))} \\
& {esc(cell(ws, "D9"))} &  & {esc(cell(ws, "F9"))} &  & \\
Yes (4-5) & {esc(cell(ws, "D10"))} &  & {esc(cell(ws, "F10"))} &  & {esc(cell(ws, "H10"))} \\
& {esc(cell(ws, "D11"))} &  & {esc(cell(ws, "F11"))} &  & \\
\addlinespace
\multicolumn{{6}}{{l}}{{\textit{{Change in financial status}}}} \\
Better &  & {esc(cell(ws, "E14"))} &  & {esc(cell(ws, "G14"))} & {esc(cell(ws, "H14"))} \\
&  & {esc(cell(ws, "E15"))} &  & {esc(cell(ws, "G15"))} & \\
No change &  & {esc(cell(ws, "E16"))} &  & {esc(cell(ws, "G16"))} & {esc(cell(ws, "H16"))} \\
&  & {esc(cell(ws, "E17"))} &  & {esc(cell(ws, "G17"))} & \\
Worse &  & {esc(cell(ws, "E18"))} &  & {esc(cell(ws, "G18"))} & {esc(cell(ws, "H18"))} \\
&  & {esc(cell(ws, "E19"))} &  & {esc(cell(ws, "G19"))} & \\
\midrule
Observations & {esc(cell(ws, "D21"))} & {esc(cell(ws, "E21"))} & {esc(cell(ws, "F21"))} & {esc(cell(ws, "G21"))} & \\
""".strip()

    header = r"& Bame & Bame & Non-Bame & Non-Bame & p-value \\"
    notes = esc(cell(ws, "A23")) + " " + esc(cell(ws, "A24"))
    tex_text = table_template(title, r"p{5.2cm}ccccc", header, body, notes)
    write_text(tex, tex_text)


# ---------- housing (updated: no home ownership) ----------
def convert_housing():
    xlsx = out_dir / "housing_tables" / "Table_BAME_housing.xlsx"
    tex = out_dir / "housing_tables" / "Table_BAME_housing.tex"
    ws = load_ws(xlsx)

    title = esc(cell(ws, "A2"))

    body = rf"""
\multicolumn{{6}}{{l}}{{\textit{{Bedroom-per-person ratio at baseline}}}} \\
$<$1 per person & {esc(cell(ws, "D8"))} &  & {esc(cell(ws, "F8"))} &  & {esc(cell(ws, "H8"))} \\
& {esc(cell(ws, "D9"))} &  & {esc(cell(ws, "F9"))} &  & \\
1-$<$2 per person & {esc(cell(ws, "D10"))} &  & {esc(cell(ws, "F10"))} &  & {esc(cell(ws, "H10"))} \\
& {esc(cell(ws, "D11"))} &  & {esc(cell(ws, "F11"))} &  & \\
$\geq$2 per person & {esc(cell(ws, "D12"))} &  & {esc(cell(ws, "F12"))} &  & {esc(cell(ws, "H12"))} \\
& {esc(cell(ws, "D13"))} &  & {esc(cell(ws, "F13"))} &  & \\
\addlinespace
\multicolumn{{6}}{{l}}{{\textit{{Other rooms at baseline}}}} \\
0-1 rooms &  & {esc(cell(ws, "E16"))} &  & {esc(cell(ws, "G16"))} & {esc(cell(ws, "H16"))} \\
&  & {esc(cell(ws, "E17"))} &  & {esc(cell(ws, "G17"))} & \\
2-3 rooms &  & {esc(cell(ws, "E18"))} &  & {esc(cell(ws, "G18"))} & {esc(cell(ws, "H18"))} \\
&  & {esc(cell(ws, "E19"))} &  & {esc(cell(ws, "G19"))} & \\
4+ rooms &  & {esc(cell(ws, "E20"))} &  & {esc(cell(ws, "G20"))} & {esc(cell(ws, "H20"))} \\
&  & {esc(cell(ws, "E21"))} &  & {esc(cell(ws, "G21"))} & \\
\midrule
Observations & {esc(cell(ws, "D23"))} & {esc(cell(ws, "E23"))} & {esc(cell(ws, "F23"))} & {esc(cell(ws, "G23"))} & \\
""".strip()

    header = r"& Bame & Bame & Non-Bame & Non-Bame & p-value \\"
    notes = esc(cell(ws, "A25")) + " " + esc(cell(ws, "A26"))
    tex_text = table_template(title, r"p{5.3cm}ccccc", header, body, notes)
    write_text(tex, tex_text)


# ---------- health A (updated: no health services change) ----------
def convert_health_a():
    xlsx = out_dir / "health_tables" / "Table_BAME_health_A.xlsx"
    tex = out_dir / "health_tables" / "Table_BAME_health_A.tex"
    ws = load_ws(xlsx)

    title = esc(cell(ws, "A2"))

    body = rf"""
\multicolumn{{8}}{{l}}{{\textit{{COVID symptoms}}}} \\
No & {esc(cell(ws, "D8"))} &  &  & {esc(cell(ws, "G8"))} &  &  & {esc(cell(ws, "J8"))} \\
& {esc(cell(ws, "D9"))} &  &  & {esc(cell(ws, "G9"))} &  &  & \\
Yes & {esc(cell(ws, "D10"))} &  &  & {esc(cell(ws, "G10"))} &  &  & {esc(cell(ws, "J10"))} \\
& {esc(cell(ws, "D11"))} &  &  & {esc(cell(ws, "G11"))} &  &  & \\
\addlinespace
\multicolumn{{8}}{{l}}{{\textit{{Respiratory chronic condition}}}} \\
No &  & {esc(cell(ws, "E14"))} &  &  & {esc(cell(ws, "H14"))} &  & {esc(cell(ws, "J14"))} \\
&  & {esc(cell(ws, "E15"))} &  &  & {esc(cell(ws, "H15"))} &  & \\
Yes &  & {esc(cell(ws, "E16"))} &  &  & {esc(cell(ws, "H16"))} &  & {esc(cell(ws, "J16"))} \\
&  & {esc(cell(ws, "E17"))} &  &  & {esc(cell(ws, "H17"))} &  & \\
\addlinespace
\multicolumn{{8}}{{l}}{{\textit{{Cardiovascular chronic condition}}}} \\
No &  &  & {esc(cell(ws, "F20"))} &  &  & {esc(cell(ws, "I20"))} & {esc(cell(ws, "J20"))} \\
&  &  & {esc(cell(ws, "F21"))} &  &  & {esc(cell(ws, "I21"))} & \\
Yes &  &  & {esc(cell(ws, "F22"))} &  &  & {esc(cell(ws, "I22"))} & {esc(cell(ws, "J22"))} \\
&  &  & {esc(cell(ws, "F23"))} &  &  & {esc(cell(ws, "I23"))} & \\
\midrule
Observations & {esc(cell(ws, "D25"))} & {esc(cell(ws, "E25"))} & {esc(cell(ws, "F25"))} & {esc(cell(ws, "G25"))} & {esc(cell(ws, "H25"))} & {esc(cell(ws, "I25"))} & \\
""".strip()

    header = r"& Bame & Bame & Bame & Non-Bame & Non-Bame & Non-Bame & p-value \\"
    notes = esc(cell(ws, "A27")) + " " + esc(cell(ws, "A28"))
    tex_text = table_template(title, r"p{5.4cm}ccccccc", header, body, notes)
    write_text(tex, tex_text)


# ---------- health B ----------
def convert_health_b():
    xlsx = out_dir / "health_tables" / "Table_BAME_health_B.xlsx"
    tex = out_dir / "health_tables" / "Table_BAME_health_B.tex"
    ws = load_ws(xlsx)

    title = esc(cell(ws, "A2"))

    body = rf"""
\multicolumn{{8}}{{l}}{{\textit{{Endocrine chronic condition}}}} \\
No & {esc(cell(ws, "D8"))} &  &  & {esc(cell(ws, "G8"))} &  &  & {esc(cell(ws, "J8"))} \\
& {esc(cell(ws, "D9"))} &  &  & {esc(cell(ws, "G9"))} &  &  & \\
Yes & {esc(cell(ws, "D10"))} &  &  & {esc(cell(ws, "G10"))} &  &  & {esc(cell(ws, "J10"))} \\
& {esc(cell(ws, "D11"))} &  &  & {esc(cell(ws, "G11"))} &  &  & \\
\addlinespace
\multicolumn{{8}}{{l}}{{\textit{{Arthritis}}}} \\
No &  & {esc(cell(ws, "E14"))} &  &  & {esc(cell(ws, "H14"))} &  & {esc(cell(ws, "J14"))} \\
&  & {esc(cell(ws, "E15"))} &  &  & {esc(cell(ws, "H15"))} &  & \\
Yes &  & {esc(cell(ws, "E16"))} &  &  & {esc(cell(ws, "H16"))} &  & {esc(cell(ws, "J16"))} \\
&  & {esc(cell(ws, "E17"))} &  &  & {esc(cell(ws, "H17"))} &  & \\
\addlinespace
\multicolumn{{8}}{{l}}{{\textit{{Other chronic condition}}}} \\
No &  &  & {esc(cell(ws, "F20"))} &  &  & {esc(cell(ws, "I20"))} & {esc(cell(ws, "J20"))} \\
&  &  & {esc(cell(ws, "F21"))} &  &  & {esc(cell(ws, "I21"))} & \\
Yes &  &  & {esc(cell(ws, "F22"))} &  &  & {esc(cell(ws, "I22"))} & {esc(cell(ws, "J22"))} \\
&  &  & {esc(cell(ws, "F23"))} &  &  & {esc(cell(ws, "I23"))} & \\
\midrule
Observations & {esc(cell(ws, "D25"))} & {esc(cell(ws, "E25"))} & {esc(cell(ws, "F25"))} & {esc(cell(ws, "G25"))} & {esc(cell(ws, "H25"))} & {esc(cell(ws, "I25"))} & \\
""".strip()

    header = r"& Bame & Bame & Bame & Non-Bame & Non-Bame & Non-Bame & p-value \\"
    notes = esc(cell(ws, "A27")) + " " + esc(cell(ws, "A28"))
    tex_text = table_template(title, r"p{5.4cm}ccccccc", header, body, notes)
    write_text(tex, tex_text)


# ---------- household composition ----------
def convert_hhcomp():
    xlsx = out_dir / "hhcomp_tables" / "Table_BAME_hhcomp.xlsx"
    tex = out_dir / "hhcomp_tables" / "Table_BAME_hhcomp.tex"
    ws = load_ws(xlsx)

    title = esc(cell(ws, "A2"))

    body = rf"""
\multicolumn{{6}}{{l}}{{\textit{{Household occupancy at baseline}}}} \\
Multi-occupancy & {esc(cell(ws, "D8"))} &  & {esc(cell(ws, "F8"))} &  & {esc(cell(ws, "H8"))} \\
& {esc(cell(ws, "D9"))} &  & {esc(cell(ws, "F9"))} &  & \\
Single & {esc(cell(ws, "D10"))} &  & {esc(cell(ws, "F10"))} &  & {esc(cell(ws, "H10"))} \\
& {esc(cell(ws, "D11"))} &  & {esc(cell(ws, "F11"))} &  & \\
\addlinespace
\multicolumn{{6}}{{l}}{{\textit{{Dependent children at baseline}}}} \\
No kids &  & {esc(cell(ws, "E14"))} &  & {esc(cell(ws, "G14"))} & {esc(cell(ws, "H14"))} \\
&  & {esc(cell(ws, "E15"))} &  & {esc(cell(ws, "G15"))} & \\
1-2 dep kids &  & {esc(cell(ws, "E16"))} &  & {esc(cell(ws, "G16"))} & {esc(cell(ws, "H16"))} \\
&  & {esc(cell(ws, "E17"))} &  & {esc(cell(ws, "G17"))} & \\
$\geq$3 dep kids &  & {esc(cell(ws, "E18"))} &  & {esc(cell(ws, "G18"))} & {esc(cell(ws, "H18"))} \\
&  & {esc(cell(ws, "E19"))} &  & {esc(cell(ws, "G19"))} & \\
\midrule
Observations & {esc(cell(ws, "D21"))} & {esc(cell(ws, "E21"))} & {esc(cell(ws, "F21"))} & {esc(cell(ws, "G21"))} & \\
""".strip()

    header = r"& Bame & Bame & Non-Bame & Non-Bame & p-value \\"
    notes = esc(cell(ws, "A23")) + " " + esc(cell(ws, "A24"))
    tex_text = table_template(title, r"p{5.2cm}ccccc", header, body, notes)
    write_text(tex, tex_text)


# ---------- demographics ----------
def convert_demo():
    xlsx = out_dir / "demo_tables" / "Table_BAME_demographics.xlsx"
    tex = out_dir / "demo_tables" / "Table_BAME_demographics.tex"
    ws = load_ws(xlsx)

    title = esc(cell(ws, "A2"))

    body = rf"""
\multicolumn{{6}}{{l}}{{\textit{{Sex}}}} \\
Male & {esc(cell(ws, "D8"))} &  & {esc(cell(ws, "F8"))} &  & {esc(cell(ws, "H8"))} \\
& {esc(cell(ws, "D9"))} &  & {esc(cell(ws, "F9"))} &  & \\
Female & {esc(cell(ws, "D10"))} &  & {esc(cell(ws, "F10"))} &  & {esc(cell(ws, "H10"))} \\
& {esc(cell(ws, "D11"))} &  & {esc(cell(ws, "F11"))} &  & \\
\addlinespace
\multicolumn{{6}}{{l}}{{\textit{{Age group}}}} \\
16-29 &  & {esc(cell(ws, "E14"))} &  & {esc(cell(ws, "G14"))} & {esc(cell(ws, "H14"))} \\
&  & {esc(cell(ws, "E15"))} &  & {esc(cell(ws, "G15"))} & \\
30-49 &  & {esc(cell(ws, "E16"))} &  & {esc(cell(ws, "G16"))} & {esc(cell(ws, "H16"))} \\
&  & {esc(cell(ws, "E17"))} &  & {esc(cell(ws, "G17"))} & \\
50-69 &  & {esc(cell(ws, "E18"))} &  & {esc(cell(ws, "G18"))} & {esc(cell(ws, "H18"))} \\
&  & {esc(cell(ws, "E19"))} &  & {esc(cell(ws, "G19"))} & \\
70+ &  & {esc(cell(ws, "E20"))} &  & {esc(cell(ws, "G20"))} & {esc(cell(ws, "H20"))} \\
&  & {esc(cell(ws, "E21"))} &  & {esc(cell(ws, "G21"))} & \\
\midrule
Observations & {esc(cell(ws, "D23"))} & {esc(cell(ws, "E23"))} & {esc(cell(ws, "F23"))} & {esc(cell(ws, "G23"))} & \\
""".strip()

    header = r"& Bame & Bame & Non-Bame & Non-Bame & p-value \\"
    notes = esc(cell(ws, "A25")) + " " + esc(cell(ws, "A26"))
    tex_text = table_template(title, r"p{4.8cm}ccccc", header, body, notes)
    write_text(tex, tex_text)


# ---------- master include file ----------
def build_master_include():
    text = r"""
% Required packages in your preamble:
% \usepackage{booktabs}
% \usepackage{adjustbox}

\input{employment_tables/Table_BAME_custom.tex}

\input{financial_tables/Table_BAME_financial.tex}

\input{housing_tables/Table_BAME_housing.tex}

\input{health_tables/Table_BAME_health_A.tex}
\input{health_tables/Table_BAME_health_B.tex}

\input{hhcomp_tables/Table_BAME_hhcomp.tex}

\input{demo_tables/Table_BAME_demographics.tex}
""".strip()

    write_text(out_dir / "all_tables_input_example.tex", text)


def main():
    convert_employment()
    convert_financial()
    convert_housing()
    convert_health_a()
    convert_health_b()
    convert_hhcomp()
    convert_demo()
    build_master_include()


if __name__ == "__main__":
    main()
```
