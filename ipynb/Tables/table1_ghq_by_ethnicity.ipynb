```python
from pathlib import Path
import pandas as pd
import re

input_csv = Path(r"/Users/lishixue/Documents/Master thesis/Statafile/Thesis2026/out/Tables/descriptive table/ethnicity_mean_d_ghq_likert.csv")
output_tex = Path(r"/Users/lishixue/Documents/Master thesis/Statafile/Thesis2026/out/Tables/descriptive table/ethnicity_mean_d_ghq_likert.tex")

def latex_escape(text):
    if pd.isna(text):
        return ""
    text = str(text)
    replacements = {
        "\\": r"\textbackslash{}",
        "&": r"\&",
        "%": r"\%",
        "$": r"\$",
        "#": r"\#",
        "_": r"\_",
        "{": r"\{",
        "}": r"\}",
        "~": r"\textasciitilde{}",
        "^": r"\textasciicircum{}",
    }
    pattern = re.compile("|".join(re.escape(k) for k in replacements))
    return pattern.sub(lambda m: replacements[m.group(0)], text)

df = pd.read_csv(input_csv)

cols = [
    "ethn_str",
    "mean_2019_ghq_likert",
    "mean_CA_ghq_likert",
    "mean_d_ghq_likert",
    "sd_d_ghq_likert",
    "n",
]
missing = [c for c in cols if c not in df.columns]
if missing:
    raise ValueError(f"CSV 缺少欄位: {missing}")

df = df[cols].copy()

num_cols = [
    "mean_2019_ghq_likert",
    "mean_CA_ghq_likert",
    "mean_d_ghq_likert",
    "sd_d_ghq_likert",
    "n",
]
for c in num_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df = df.sort_values("mean_d_ghq_likert").reset_index(drop=True)
df["ethn_str"] = df["ethn_str"].map(latex_escape)

lines = []
lines.append(r"\begin{table}[htbp]")
lines.append(r"\centering")
lines.append(r"\caption{Descriptive Statistics of GHQ Likert Change by Ethnicity}")
lines.append(r"\label{tab:desc_ethnicity_ghq}")
lines.append(r"\begin{threeparttable}")
lines.append(r"\footnotesize")
lines.append(r"\begin{tabular*}{\textwidth}{@{\extracolsep{\fill}}p{5.2cm}rrrrr}")
lines.append(r"\toprule")
lines.append(r"Ethnicity & Pre-COVID & CA & Change & SD & N \\")
lines.append(r"\midrule")

for _, row in df.iterrows():
    ethn = row["ethn_str"]
    m2019 = "" if pd.isna(row["mean_2019_ghq_likert"]) else f'{row["mean_2019_ghq_likert"]:.2f}'
    mca   = "" if pd.isna(row["mean_CA_ghq_likert"]) else f'{row["mean_CA_ghq_likert"]:.2f}'
    md    = "" if pd.isna(row["mean_d_ghq_likert"]) else f'{row["mean_d_ghq_likert"]:.2f}'
    sd    = "" if pd.isna(row["sd_d_ghq_likert"]) else f'{row["sd_d_ghq_likert"]:.2f}'
    n     = "" if pd.isna(row["n"]) else f'{int(round(row["n"])):,}'
    lines.append(f"{ethn} & {m2019} & {mca} & {md} & {sd} & {n} \\\\")

lines.append(r"\bottomrule")
lines.append(r"\end{tabular*}")
lines.append(r"\begin{tablenotes}[flushleft]")
lines.append(r"\footnotesize")
lines.append(
    r"\item Notes: This table reports ethnicity-specific descriptive statistics for the April-adjusted, standardized, and inverted GHQ Likert measure. Pre-COVID and Post-COVID correspond to the weighted mean values in 2019 and the first wave of the COVID survey, respectively. Change is defined as the individual-level difference between the post-COVID and 2019. SD denotes the unweighted standard deviation of the change, and N denotes the unweighted sample size. Higher values indicate better mental health."
)
lines.append(r"\end{tablenotes}")
lines.append(r"\end{threeparttable}")
lines.append(r"\end{table}")

latex_table = "\n".join(lines)
output_tex.write_text(latex_table, encoding="utf-8")

print(f"LaTeX table saved to: {output_tex}")
print()
print(latex_table)
```
